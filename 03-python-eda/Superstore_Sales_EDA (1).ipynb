```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_excel(r"D:\teks\OCTOBER PROJECT\Sample - Superstore.xlsx")

print("Dataset loaded successfully!")
print("Rows and columns:", df.shape)

df.head()
```

```python
print("Dataset dimensions:", df.shape)

print("\nColumn names:")
print(df.columns)

print("\nDataset information:")
df.info()
```

```python
print("Missing values in each column:")
print(df.isnull().sum())

print("\nNumber of duplicate rows:")
print(df.duplicated().sum())
```

```python
df["Order Date"] = pd.to_datetime(df["Order Date"])
df["Ship Date"] = pd.to_datetime(df["Ship Date"])

print(df[["Order Date", "Ship Date"]].dtypes)
```

```python
total_sales = df["Sales"].sum()
total_profit = df["Profit"].sum()
total_orders = df["Order ID"].nunique()
profit_margin = (total_profit / total_sales) * 100

print("Total Sales:", round(total_sales, 2))
print("Total Profit:", round(total_profit, 2))
print("Total Orders:", total_orders)
print("Profit Margin:", round(profit_margin, 2), "%")
```

```python
print(df[["Order Date", "Ship Date"]].head(20))

print("\nDate data types:")
print(df[["Order Date", "Ship Date"]].dtypes)

print("\nOrder Date range:")
print(df["Order Date"].min(), "to", df["Order Date"].max())
```

```python
print(df["Order Date"].head(10).to_list())   #Check the original date format
print(df["Ship Date"].head(10).to_list())
```

```python
print("Ship dates before order dates:")    #dates columns are complicated .....in power bi added new columns as reaarranged dates...so leavind dates column for now
print((df["Ship Date"] < df["Order Date"]).sum())
```

```python
total_sales = df["Sales"].sum()
total_profit = df["Profit"].sum()
total_orders = df["Order ID"].nunique()
profit_margin = (total_profit / total_sales) * 100

print("Total Sales:", round(total_sales, 2))
print("Total Profit:", round(total_profit, 2))
print("Total Orders:", total_orders)
print("Profit Margin:", round(profit_margin, 2), "%")
```

```python
category_analysis = df.groupby("Category")[["Sales", "Profit"]].sum()

category_analysis = category_analysis.sort_values(
    by="Sales", ascending=False
)

print(category_analysis.round(2))
```

```python
subcategory_analysis = df.groupby("Sub-Category")[["Sales", "Profit"]].sum()

subcategory_analysis = subcategory_analysis.sort_values(
    by="Profit", ascending=True
)

print(subcategory_analysis.round(2))
```

```python
subcategory_analysis["Profit"].plot(
    kind="bar",
    figsize=(12, 6),
    title="Profit by Sub-Category"
)

plt.xlabel("Sub-Category")
plt.ylabel("Profit (USD)")
plt.axhline(y=0, color="black")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
subcategory_analysis["Profit"].plot(
    kind="bar",
    figsize=(8, 4),
    title="Profit by Sub-Category"
)

plt.xlabel("Sub-Category")
plt.ylabel("Profit (USD)")
plt.axhline(y=0, color="black")
plt.xticks(rotation=45, fontsize=8)
plt.yticks(fontsize=8)
plt.tight_layout()
plt.show()
```

```python
category_analysis["Sales"].plot(
    kind="bar",
    figsize=(6, 3),
    title="Sales by Category"
)

plt.xlabel("Category")
plt.ylabel("Sales (USD)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
region_analysis = df.groupby("Region")[["Sales", "Profit"]].sum()

region_analysis = region_analysis.sort_values(
    by="Profit", ascending=False
)

print(region_analysis.round(2))
```

```python
region_analysis["Profit"].plot(
    kind="bar",
    figsize=(6, 3),
    title="Profit by Region"
)

plt.xlabel("Region")
plt.ylabel("Profit (USD)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
region_analysis[["Sales", "Profit"]].plot(
    kind="bar",
    figsize=(7, 4),
    title="Sales vs Profit by Region"
)

plt.xlabel("Region")
plt.ylabel("Amount (USD)")
plt.xticks(rotation=0)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
region_analysis[["Sales", "Profit"]].plot(
    kind="bar",
    figsize=(7, 4),
    title="Sales vs Profit by Region",
    color=["purple", "gold"]
)

plt.xlabel("Region")
plt.ylabel("Amount (USD)")
plt.xticks(rotation=0)
plt.legend()
plt.tight_layout()
plt.show()
```

```python

fig, axes = plt.subplots(3, 2, figsize=(14, 12))

# 1. Sales by Category
category_analysis["Sales"].plot(
    kind="bar", ax=axes[0, 0], title="Sales by Category",
    color="royalblue"
)

# 2. Profit by Sub-Category
subcategory_analysis["Profit"].plot(
    kind="bar", ax=axes[0, 1], title="Profit by Sub-Category",
    color="orange"
)

# 3. Profit by Region
region_analysis["Profit"].plot(
    kind="bar", ax=axes[1, 0], title="Profit by Region",
    color="seagreen"
)

# 4. Sales vs Profit by Region
region_analysis[["Sales", "Profit"]].plot(
    kind="bar", ax=axes[1, 1],
    title="Sales vs Profit by Region",
    color=["royalblue", "orange"]
)

# 5. Sales Contribution by Category (Pie Chart)
category_analysis["Sales"].plot(
    kind="pie", ax=axes[2, 0],
    autopct="%1.1f%%",
    startangle=90,
    title="Sales Contribution by Category",
    ylabel="",
    colormap="Set3"
)

axes[0, 0].set_ylabel("Sales (USD)")
axes[0, 1].set_ylabel("Profit (USD)")
axes[1, 0].set_ylabel("Profit (USD)")
axes[1, 1].set_ylabel("Amount (USD)")

axes[0, 1].tick_params(axis="x", labelrotation=45)
axes[1, 0].tick_params(axis="x", labelrotation=0)
axes[1, 1].tick_params(axis="x", labelrotation=0)

# Hide the unused sixth subplot
axes[2, 1].axis("off")

for ax in [axes[0, 0], axes[0, 1],
           axes[1, 0], axes[1, 1]]:
    ax.axhline(y=0, color="black", linewidth=0.6)
    ax.tick_params(axis="both", labelsize=8)

axes[2, 0].set_aspect("equal")

plt.suptitle(
    "Superstore Sales & Profitability Analysis",
    fontsize=16
)
plt.tight_layout()
plt.show()
```
